```python
# understand the date
# Convert the "date" column to datetime format
sales["date"] = pd.to_datetime(sales["date"])
```

```python
# find the beginning date
# Check the minimum date
sales["date"].min()
```

```python
# find the end date
# Check the maximum date
sales["date"].max()
```

```python
# together
print("Start:", sales["date"].min())
print("End:", sales["date"].max())
```

```python
# Find the number of unique products/items
sales["item_id"].nunique()
```

```python
# Find the number of unique stores
sales["store_id"].nunique()
```

```python
# Find the total quantity sold
sales["quantity"].sum()
```

```python
# Calculate daily sales
daily_sales = sales.groupby("date")["sum_total"].sum()
```

```python
# display the daily sales in table format
daily_sales.reset_index()
```

```python
# Plotting the daily sales trend
daily_sales.plot(figsize=(12, 5))

plt.title("Daily Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")

plt.show()
```
